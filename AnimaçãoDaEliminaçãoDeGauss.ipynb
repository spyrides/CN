```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

A_g = np.array([[2., 1., -1.],
                [-3., -1., 2.],
                [-2., 1., 2.]])
b_g = np.array([8., -11., -3.])
M_g = np.column_stack((A_g, b_g))

estados = []

estados.append({
    'M': M_g.copy(), 'pivo': None, 'linha_alvo': None,
    'msg': "Matriz Aumentada Inicial"
})

pivo_1 = (0, 0)
estados.append({
    'M': M_g.copy(), 'pivo': pivo_1, 'linha_alvo': None,
    'msg': f"Selecionando o Pivô 1: M[0,0] = {M_g[0,0]:.2f}"
})

m21 = M_g[1,0]/M_g[0,0]
estados.append({
    'M': M_g.copy(), 'pivo': pivo_1, 'linha_alvo': 1,
    'msg': f"Multiplicador m21 = {m21:.2f}\nL2 \u2192 L2 - ({m21:.2f})*L1"
})
M_g[1,:] -= m21 * M_g[0,:]
estados.append({
    'M': M_g.copy(), 'pivo': pivo_1, 'linha_alvo': 1,
    'msg': "Linha 2 atualizada com sucesso!"
})

m31 = M_g[2,0]/M_g[0,0]
estados.append({
    'M': M_g.copy(), 'pivo': pivo_1, 'linha_alvo': 2,
    'msg': f"Multiplicador m31 = {m31:.2f}\nL3 \u2192 L3 - ({m31:.2f})*L1"
})
M_g[2,:] -= m31 * M_g[0,:]
estados.append({
    'M': M_g.copy(), 'pivo': pivo_1, 'linha_alvo': 2,
    'msg': "Linha 3 atualizada com sucesso!"
})


pivo_2 = (1, 1)
estados.append({
    'M': M_g.copy(), 'pivo': pivo_2, 'linha_alvo': None,
    'msg': f"Selecionando o Pivô 2: M[1,1] = {M_g[1,1]:.2f}"
})


m32 = M_g[2,1]/M_g[1,1]
estados.append({
    'M': M_g.copy(), 'pivo': pivo_2, 'linha_alvo': 2,
    'msg': f"Multiplicador m32 = {m32:.2f}\nL3 \u2192 L3 - ({m32:.2f})*L2"
})
M_g[2,:] -= m32 * M_g[1,:]

# Estado Final
estados.append({
    'M': M_g.copy(), 'pivo': None, 'linha_alvo': None,
    'msg': "Matriz Final (Escalonada)!"
})

for _ in range(3):
    estados.append(estados[-1])

fig, ax = plt.subplots(figsize=(8, 5))

def desenhar_estado(frame):
    ax.clear()
    ax.axis('off') # Esconder eixos

    estado = estados[frame]
    M = estado['M']
    msg = estado['msg']
    pivo = estado['pivo']
    linha_alvo = estado['linha_alvo']

    # Título e Mensagens
    ax.text(0.5, 0.95, "Animação: Eliminação de Gauss", ha='center', va='center',
            fontsize=16, weight='bold', transform=ax.transAxes)
    ax.text(0.5, 0.82, msg, ha='center', va='center',
            fontsize=13, color='#333399', transform=ax.transAxes)

    dados_tabela = [[f"{val:.2f}" for val in linha] for linha in M]

    tabela = ax.table(cellText=dados_tabela, loc='center', cellLoc='center',
                      bbox=[0.1, 0.1, 0.8, 0.6])
    tabela.auto_set_font_size(False)
    tabela.set_fontsize(14)

    # Aplicar cores para destacar pivôs e linhas alteradas
    for (i, j), celula in tabela.get_celld().items():
        celula.set_edgecolor('#cccccc') # bordas suaves

        if pivo and (i, j) == pivo:
            celula.set_facecolor('#ffcc99')
            celula.get_text().set_weight('bold')
        elif linha_alvo is not None and i == linha_alvo:
            celula.set_facecolor('#ccffcc')
        else:
            celula.set_facecolor('#f9f9f9')

        if j == 3:
            celula.set_edgecolor('black')


print("Gerando a animação, aguarde...")
animacao = FuncAnimation(fig, desenhar_estado, frames=len(estados), interval=1500)


animacao.save('eliminacao_gauss.gif', writer=PillowWriter(fps=0.6))
print("GIF gerado com sucesso: 'eliminacao_gauss.gif'")
plt.close()
```
