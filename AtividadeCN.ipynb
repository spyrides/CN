# PARTE A - Truncamento e Arredondamento

1. calcule a aproximação por truncamento e por arredondamento;

```python
import math
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

print("Ambiente preparado com sucesso :)) yay \n")

x1_referencia = 3.14159265
x2_referencia = 98.76546
x3_referencia = 0.00498765

casas_decimais = [0,1,2,3,4]

resultados = np.zeros((6,5))
erro_abso = np.zeros((6,5))
erro_relat = np.zeros((6,5))
erro_percent = np.zeros((6,5))

def erro_absoluto(ref,aprox):
  return abs(ref - aprox)

def erro_relativo(ref,erro_abs):
  return erro_abs / abs(ref)

def erro_percentual(erro_rel):
  return 100 * erro_rel

pos = 0
linha = 0

for casa in casas_decimais:
  print(f"\nPara {casa} casas decimais:")

  print("TRUNCAMENTO: ")

  # x1 ---------------------------
  res = math.trunc(x1_referencia * (10 ** casa)) / 10 ** casa
  resultados[linha,pos] = res
  print(f"x1 = {res}")

  erro = erro_absoluto(x1_referencia,res)
  erro_abso[linha,pos] = erro
  print(f"   erro absoluto = {erro}")

  erro = erro_relativo(x1_referencia,erro)
  erro_relat[linha,pos] = erro
  print(f"   erro relativo = {erro}")

  erro = erro_percentual(erro)
  erro_percent[linha,pos] = erro
  print(f"   erro percentual = {erro}")

  linha = linha + 1

   # x2 ---------------------------
  res = math.trunc(x2_referencia * (10 ** casa)) / 10 ** casa
  resultados[linha,pos] = res
  print(f"x2 = {res}")

  erro = erro_absoluto(x2_referencia,res)
  erro_abso[linha,pos] = erro
  print(f"   erro absoluto = {erro}")

  erro = erro_relativo(x2_referencia,erro)
  erro_relat[linha,pos] = erro
  print(f"   erro relativo = {erro}")

  erro = erro_percentual(erro)
  erro_percent[linha,pos] = erro
  print(f"   erro percentual = {erro}")

  linha = linha + 1


   # x3 ---------------------------
  res = math.trunc(x3_referencia * (10 ** casa)) / 10 ** casa
  resultados[linha,pos] = res
  print(f"x3 = {res}\n")

  erro = erro_absoluto(x3_referencia,res)
  erro_abso[linha,pos] = erro
  print(f"   erro absoluto = {erro}")

  erro = erro_relativo(x3_referencia,erro)
  erro_relat[linha,pos] = erro
  print(f"   erro relativo = {erro}")

  erro = erro_percentual(erro)
  erro_percent[linha,pos] = erro
  print(f"   erro percentual = {erro}")

  linha = linha + 1


  print("ARREDONDAMENTO: ")

   # x1 ---------------------------
  res = round(x1_referencia, casa)
  resultados[linha,pos] = res
  print(f"x1 = {res}")

  erro = erro_absoluto(x1_referencia,res)
  erro_abso[linha,pos] = erro
  print(f"   erro absoluto = {erro}")

  erro = erro_relativo(x1_referencia,erro)
  erro_relat[linha,pos] = erro
  print(f"   erro relativo = {erro}")

  erro = erro_percentual(erro)
  erro_percent[linha,pos] = erro
  print(f"   erro percentual = {erro}")

  linha = linha + 1

   # x2 ---------------------------
  res = round(x2_referencia, casa)
  resultados[linha,pos] = res
  print(f"x2 = {res}")

  erro = erro_absoluto(x2_referencia,res)
  erro_abso[linha,pos] = erro
  print(f"   erro absoluto = {erro}")

  erro = erro_relativo(x2_referencia,erro)
  erro_relat[linha,pos] = erro
  print(f"   erro relativo = {erro}")

  erro = erro_percentual(erro)
  erro_percent[linha,pos] = erro
  print(f"   erro percentual = {erro}")

  linha = linha + 1

   # x3 ---------------------------
  res = round(x3_referencia, casa)
  resultados[linha,pos] = res
  print(f"x3 = {res}")

  erro = erro_absoluto(x3_referencia,res)
  erro_abso[linha,pos] = erro
  print(f"   erro absoluto = {erro}")

  erro = erro_relativo(x3_referencia,erro)
  erro_relat[linha,pos] = erro
  print(f"   erro relativo = {erro}")

  erro = erro_percentual(erro)
  erro_percent[linha,pos] = erro
  print(f"   erro percentual = {erro}")

  linha = 0
  pos = pos + 1


### 4. Gráfico em escala logarítmica

# criando o grafico em escala log para o erro absoluto
plt.figure(figsize=(10, 6))

nomes_trunc = ['x1 trunc', 'x2 trunc', 'x3 trunc']
nomes_arred = ['x1 arred', 'x2 arred', 'x3 arred']

# plotando as linhas de truncamento
for i in range(3):
    plt.plot(casas_decimais, erro_abso[i, :], marker='o', linestyle='--', label=nomes_trunc[i])

# plotando as linhas de arredondamento
for i in range(3):
    plt.plot(casas_decimais, erro_abso[i+3, :], marker='s', linestyle='-', label=nomes_arred[i])

plt.yscale('log')
plt.xlabel('Número de casas decimais (n)')
plt.ylabel('Erro Absoluto (escala log)')
plt.title('Erro Absoluto em função do número de casas decimais')
plt.legend()
plt.show()
```

### 5. Arredondar sempre gera erro menor ou igual ao truncamento?
**Resposta:** Sim. O arredondamento procura o número representável mais próximo do valor real na casa decimal escolhida, o que matematicamente garante a menor distância (menor erro absoluto) em relação ao original. O truncamento simplesmente descarta os números à direita, ignorando qual é o valor mais próximo. Olhando as tabelas acima, percebe-se que o erro de arredondamento nunca superou o de truncamento; na maioria das vezes, foi menor, e em alguns casos, igual (quando o dígito descartado era pequeno).

# PARTE B - Medidas de pressão arterial

```python
# B.1 Representacao das medidas
medidas_originais = np.array([121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5])
medidas_arredondadas = np.round(medidas_originais)
medidas_truncadas = np.trunc(medidas_originais)

media_ref = np.mean(medidas_originais) # referencia

# funcao simples p/ calcular e mostrar as estatisticas
def exibir_estatisticas(nome, dados):
    media = np.mean(dados)
    minimo = np.min(dados)
    maximo = np.max(dados)
    amplitude = maximo - minimo
    desvio = np.std(dados, ddof=1) # amostral
    print(f"--- {nome} ---")
    print(f"Média: {media:.4f} | Mín: {minimo} | Máx: {maximo} | Ampl: {amplitude:.4f} | Desvio: {desvio:.4f}")
    return media

print("--- ESTATÍSTICAS ---")
media_orig = exibir_estatisticas("Originais", medidas_originais)
media_arred = exibir_estatisticas("Arredondados", medidas_arredondadas)
media_trunc = exibir_estatisticas("Truncados", medidas_truncadas)

# calculando erros das medias
erro_abs_arred = erro_absoluto(media_ref, media_arred)
erro_rel_arred = erro_relativo(media_ref, erro_abs_arred)
erro_perc_arred = erro_percentual(erro_rel_arred)

erro_abs_trunc = erro_absoluto(media_ref, media_trunc)
erro_rel_trunc = erro_relativo(media_ref, erro_abs_trunc)
erro_perc_trunc = erro_percentual(erro_rel_trunc)

print("\n--- ERROS DAS MÉDIAS ---")
print(f"Arredondada -> Ea: {erro_abs_arred:.4f} | Er: {erro_rel_arred:.6f} | E%: {erro_perc_arred:.4f}%")
print(f"Truncada    -> Ea: {erro_abs_trunc:.4f} | Er: {erro_rel_trunc:.6f} | E%: {erro_perc_trunc:.4f}%")

# B.2 Gráficos Obrigatórios
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# 1. grafico das 10 medicoes
axs[0, 0].plot(medidas_originais, marker='o', label='Originais')
axs[0, 0].plot(medidas_arredondadas, marker='s', label='Arredondados')
axs[0, 0].plot(medidas_truncadas, marker='^', label='Truncados')
axs[0, 0].axhline(media_ref, color='r', linestyle='--', label='Média Ref')
axs[0, 0].set_title("Dez Medições de Pressão")
axs[0, 0].legend()

# 2. barras dos erros absolutos
axs[0, 1].bar(['Arredondada', 'Truncada'], [erro_abs_arred, erro_abs_trunc], color=['blue', 'orange'])
axs[0, 1].set_title("Erro Absoluto das Médias")

# 3. barras dos erros percentuais
axs[0, 1].bar(['Arredondada', 'Truncada'], [erro_perc_arred, erro_perc_trunc], color=['green', 'red'])
axs[1, 0].bar(['Arredondada', 'Truncada'], [erro_perc_arred, erro_perc_trunc], color=['green', 'red'])
axs[1, 0].set_title("Erro Percentual das Médias (%)")

# 4. histograma originais x arredondados
axs[1, 1].hist(medidas_originais, alpha=0.5, label='Originais', bins=5)
axs[1, 1].hist(medidas_arredondadas, alpha=0.5, label='Arredondados', bins=5)
axs[1, 1].set_title("Histograma: Originais vs Arredondados")
axs[1, 1].legend()

plt.tight_layout()
plt.show()
```

**A representação inteira é suficiente para preservar a média e a variabilidade?**
Sim. Ao arredondar os valores para o inteiro mais próximo, os desvios positivos e negativos em relação aos valores originais tendem a se anular na soma, fazendo com que a média permaneça quase inalterada (o erro percentual da média arredondada é muito baixo). O desvio padrão também permanece próximo ao original, indicando que a variabilidade da amostra não foi perdida no processo.

# PARTE C - Vazão em um tubo de pequeno diâmetro

```python
# Dados iniciais (convertendo mm para metros)
pi = np.pi
r_ref = 0.80 * 1e-3
L_ref = 0.20
mu_ref = 3.5e-3
dp_ref = 1200

# formula principal
def hagen_poiseuille(r, mu, dp, L):
    return (pi * (r**4) * dp) / (8 * mu * L)

# C.1 Precisão das variáveis
Q_ref = hagen_poiseuille(r_ref, mu_ref, dp_ref, L_ref)

# aproximações especificadas
# 1 casa em mm p/ raio: 0.80 mm já é 0.8, então as funcoes vao dar 0.8
r_arr_m = round(0.80, 1) * 1e-3
r_tru_m = math.trunc(0.80 * 10)/10 * 1e-3

mu_arr = round(mu_ref, 3)
mu_tru = math.trunc(mu_ref * 1000)/1000

dp_arr = round(dp_ref, -2) # centenas
dp_tru = math.trunc(dp_ref / 100) * 100

print("--- C.1 Precisão e Vazões (Q) ---")
print(f"Q Referência: {Q_ref}")

Q_r_arr = hagen_poiseuille(r_arr_m, mu_ref, dp_ref, L_ref)
Q_r_tru = hagen_poiseuille(r_tru_m, mu_ref, dp_ref, L_ref)
print(f"\nQ (r arred.): {Q_r_arr} | Erro %: {erro_percentual(erro_relativo(Q_ref, erro_absoluto(Q_ref, Q_r_arr))):.4f}%")
print(f"Q (r trunc.): {Q_r_tru} | Erro %: {erro_percentual(erro_relativo(Q_ref, erro_absoluto(Q_ref, Q_r_tru))):.4f}%")

Q_mu_arr = hagen_poiseuille(r_ref, mu_arr, dp_ref, L_ref)
Q_mu_tru = hagen_poiseuille(r_ref, mu_tru, dp_ref, L_ref)
print(f"Q (mu arred.): {Q_mu_arr} | Erro %: {erro_percentual(erro_relativo(Q_ref, erro_absoluto(Q_ref, Q_mu_arr))):.4f}%")
print(f"Q (mu trunc.): {Q_mu_tru} | Erro %: {erro_percentual(erro_relativo(Q_ref, erro_absoluto(Q_ref, Q_mu_tru))):.4f}%")

Q_todas = hagen_poiseuille(r_arr_m, mu_arr, dp_arr, L_ref)
print(f"Q (tudo arred.): {Q_todas} | Erro %: {erro_percentual(erro_relativo(Q_ref, erro_absoluto(Q_ref, Q_todas))):.4f}%")

# C.2 Sensibilidade
r_vals = np.linspace(0.70, 0.90, 21) # array de raios em mm
Q_lista_ref = []
Q_lista_arr = []
Q_lista_tru = []
erro_abs_r = []
erro_rel_r = []
erro_perc_r = []

for r_mm in r_vals:
    # converte p/ metro no calculo
    q_base = hagen_poiseuille(r_mm * 1e-3, mu_ref, dp_ref, L_ref)
    q_arr = hagen_poiseuille(round(r_mm, 1) * 1e-3, mu_ref, dp_ref, L_ref)
    q_tru = hagen_poiseuille(math.trunc(r_mm*10)/10 * 1e-3, mu_ref, dp_ref, L_ref)

    Q_lista_ref.append(q_base)
    Q_lista_arr.append(q_arr)
    Q_lista_tru.append(q_tru)

    e_a = erro_absoluto(q_base, q_arr)
    erro_abs_r.append(e_a)
    e_r = erro_relativo(q_base, e_a)
    erro_rel_r.append(e_r)
    erro_perc_r.append(erro_percentual(e_r))

# Graficos da Sensibilidade
fig2, axs2 = plt.subplots(2, 2, figsize=(14, 10))

axs2[0, 0].plot(r_vals, Q_lista_ref, label='Referência', color='black')
axs2[0, 0].plot(r_vals, Q_lista_arr, label='Arredondado', linestyle='--')
axs2[0, 0].set_title('Q em função do raio (r)')
axs2[0, 0].set_xlabel('Raio (mm)')
axs2[0, 0].legend()

axs2[0, 1].plot(r_vals, erro_abs_r, color='purple')
axs2[0, 1].set_title('Erro Absoluto de Q vs r')

axs2[1, 0].plot(r_vals, erro_rel_r, color='brown')
axs2[1, 0].set_title('Erro Relativo de Q vs r')

axs2[1, 1].plot(r_vals, erro_perc_r, color='red')
axs2[1, 1].set_title('Erro Percentual de Q vs r')

plt.tight_layout()
plt.show()
```

**Por que o erro no raio é amplificado na vazão?**
Na lei de Hagen-Poiseuille, o raio está elevado à quarta potência ($Q \propto r^4$). Matematicamente, a derivada indica que um pequeno erro relativo no raio é multiplicado por 4 no resultado final da vazão ($\frac{\Delta Q}{Q} \approx 4 \frac{\Delta r}{r}$). Sendo assim, qualquer pequena imprecisão, arredondamento ou desvio geométrico no diâmetro do tubo causará um impacto gigantesco (exponencial) na passagem do fluido.

# PARTE D - Escoamento em um nanocateter

```python
# D.1 Erro Geométrico
r_nano = 50e-9 # m
L_nano = 10e-6 # m
dp_nano = 5000 # Pa
# mu continua 3.5e-3

delta_rs_nm = np.arange(0.1, 5.1, 0.1) # de 0.1 a 5.0
q_bases = []
q_menos = []
q_mais = []
erros_max_abs = []
erros_max_perc = []
aprox_linear = []

for dr_nm in delta_rs_nm:
    dr_m = dr_nm * 1e-9
    r_menos = r_nano - dr_m
    r_mais = r_nano + dr_m

    q_b = hagen_poiseuille(r_nano, mu_ref, dp_nano, L_nano)
    q_men = hagen_poiseuille(r_menos, mu_ref, dp_nano, L_nano)
    q_mai = hagen_poiseuille(r_mais, mu_ref, dp_nano, L_nano)

    q_bases.append(q_b)
    q_menos.append(q_men)
    q_mais.append(q_mai)

    e_abs_men = abs(q_b - q_men)
    e_abs_mai = abs(q_b - q_mai)
    max_abs = max(e_abs_men, e_abs_mai)
    erros_max_abs.append(max_abs)

    max_perc = erro_percentual(max_abs / q_b)
    erros_max_perc.append(max_perc)

    # aprox linear: 4 * (dr / r) * 100 para percentual
    aprox_linear.append(4 * (dr_m / r_nano) * 100)

fig3, axs3 = plt.subplots(1, 3, figsize=(18, 5))
axs3[0].plot(delta_rs_nm, q_bases, label='Q(r)', color='black')
axs3[0].plot(delta_rs_nm, q_menos, label='Q(r-)')
axs3[0].plot(delta_rs_nm, q_mais, label='Q(r+)')
axs3[0].set_title("Vazões com Erro no Raio")
axs3[0].legend()

axs3[1].plot(delta_rs_nm, erros_max_abs, color='orange')
axs3[1].set_title("Erro Absoluto Máximo de Q")

axs3[2].plot(delta_rs_nm, erros_max_perc, label='Erro % Máximo', marker='o')
axs3[2].plot(delta_rs_nm, aprox_linear, label='Aprox Linear (4*dr/r)', linestyle='--')
axs3[2].set_title("Erro Percentual Máximo vs Aproximação")
axs3[2].legend()
plt.show()

# D.2 Precisao computacional
print("\n--- D.2 Precisão Computacional ---")
r32 = np.float32(50e-9)
r64 = np.float64(50e-9)

r4_32 = r32**4
r4_64 = r64**4
q_32 = (np.float32(pi) * r4_32 * np.float32(dp_nano)) / (np.float32(8) * np.float32(mu_ref) * np.float32(L_nano))
q_64 = (np.float64(pi) * r4_64 * np.float64(dp_nano)) / (np.float64(8) * np.float64(mu_ref) * np.float64(L_nano))

print(f"r^4 (float32): {r4_32} | (float64): {r4_64}")
print(f"Q (float32): {q_32} | Q (float64): {q_64}")
print(f"Diferença Relativa Q: {abs(q_64 - q_32)/q_64}")
print(f"Eps float32: {np.finfo(np.float32).eps} | Eps float64: {np.finfo(np.float64).eps}")

print("\nPerda de significância (Cancelamento):")
dr_r_vals = [1e-2, 1e-4, 1e-8, 1e-12, 1e-15]
for dr in dr_r_vals:
    # Usando precisão padrão (float64) para simular o limite
    dr_real = dr * r_nano
    q_r = hagen_poiseuille(r_nano, mu_ref, dp_nano, L_nano)
    q_r_dr = hagen_poiseuille(r_nano + dr_real, mu_ref, dp_nano, L_nano)

    delta1 = (q_r_dr - q_r) / q_r
    delta2 = (1 + dr)**4 - 1
    print(f"dr/r = {dr} -> delta1 = {delta1}, delta2 = {delta2}")
```

**Cancelamento:** À medida que a variação `dr/r` fica extremamente pequena, a equação $\delta_1 = \frac{Q(r+\Delta r)-Q(r)}{Q(r)}$ sofre **perda de significância por cancelamento subtrativo**. O computador subtrai dois números quase idênticos, perdendo muitos dígitos exatos, o que faz com que a fórmula $\delta_2$ preserve um comportamento ligeiramente mais estável antes de também sucumbir às limitações do número de bits da máquina.

**D.3 Fenômeno Físico:**
Quando diminuímos a escala para dimensões nanométricas, a razão entre a superfície e o volume do tubo se torna extremamente alta. Nesse ambiente, as forças de superfície, como a viscosidade e a interação com as paredes, predominam fortemente sobre as forças inerciais. Um cálculo em precisão altíssima (float64) entregará dezenas de casas decimais "perfeitas" para a vazão do modelo de Hagen-Poiseuille, mas fisicamente esse modelo contínuo clássico perde a validade, já que em canais tão estreitos o fluido deixa de se comportar idealmente e outros fenômenos de interface passam a governar o escoamento.

```python
# Parte E - Imagem Única de Resultados (Sintese)
fig_s, ax_s = plt.subplots(2, 2, figsize=(16, 10), dpi=100) # 1600x1000 pixels

# Card / Conclusao no topo
fig_s.suptitle("Síntese Visual - Cálculo Numérico (Nanocateter e Vazões)", fontsize=20, fontweight='bold')
texto_conclusao = (
    "Valores Principais: \n"
    f"- Pressão Média (B): {media_ref:.2f} mmHg\n"
    f"- Q ref macro (C): {Q_ref:.2e} m³/s\n"
    f"- Q ref nano (D): {q_64:.2e} m³/s\n\n"
    "Conclusão Nanocateter:\n"
    "O arredondamento se mostrou superior ou igual ao truncamento.\n"
    "Q é super sensível ao raio (Q amplifica o erro em 4x).\n"
    "Apesar da precisão do float64, o modelo macroscópico \n"
    "perde validade no nanoscale (efeitos de superfície)."
)
ax_s[0, 0].text(0.1, 0.4, texto_conclusao, fontsize=14, bbox=dict(facecolor='lightcyan', alpha=0.5))
ax_s[0, 0].axis('off')

# Erros de Trunc x Arred
ax_s[0, 1].bar(['Arredondada', 'Truncada'], [erro_perc_arred, erro_perc_trunc], color=['green', 'red'])
ax_s[0, 1].set_title("Comparação: Arredondamento x Truncamento (Erro %)")

# Sensibilidade
ax_s[1, 0].plot(r_vals, Q_lista_ref, color='blue', lw=2)
ax_s[1, 0].set_title("Sensibilidade: Vazão vs Raio")
ax_s[1, 0].set_xlabel("Raio (mm)")

# Erro no nano
ax_s[1, 1].plot(delta_rs_nm, erros_max_perc, color='red', marker='o')
ax_s[1, 1].set_title("Nanocateter: Erro % Máximo de Q (Variação de r)")
ax_s[1, 1].set_xlabel("Variação do raio (nm)")

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('sintese_resultados.png') # Salva em PNG
plt.show()

# -----------------
# 8. GIF Demonstrativo
import matplotlib.animation as animation
from matplotlib.animation import PillowWriter

fig_anim, (ax_canal, ax_curva) = plt.subplots(1, 2, figsize=(10, 5))
raios_gif = np.linspace(45, 55, 40) # 40 quadros, 45 a 55 nm
q_gif = []
raios_salvos = []

def update(frame):
    ax_canal.clear()
    ax_curva.clear()

    r_atual = raios_gif[frame]
    raios_salvos.append(r_atual)

    # 1. Secao transversal
    circulo = plt.Circle((0, 0), r_atual, color='skyblue', ec='blue')
    ax_canal.add_patch(circulo)
    ax_canal.set_xlim(-60, 60)
    ax_canal.set_ylim(-60, 60)
    ax_canal.set_aspect('equal')

    # Calculo da vazão e erro
    q_at = hagen_poiseuille(r_atual * 1e-9, mu_ref, dp_nano, L_nano)
    q_gif.append(q_at)
    q_base_50 = hagen_poiseuille(50e-9, mu_ref, dp_nano, L_nano)
    e_perc = erro_percentual(abs(q_base_50 - q_at) / q_base_50)

    ax_canal.set_title(f"Seção: r = {r_atual:.1f} nm\nErro % (ref 50nm) = {e_perc:.1f}%")

    # 2. Curva progressiva
    ax_curva.plot(raios_salvos, q_gif, color='red', lw=2)
    ax_curva.set_xlim(45, 55)
    # y limit baseado no min e max esperados (r=45 a 55)
    q_min = hagen_poiseuille(45e-9, mu_ref, dp_nano, L_nano)
    q_max = hagen_poiseuille(55e-9, mu_ref, dp_nano, L_nano)
    ax_curva.set_ylim(q_min * 0.9, q_max * 1.1)
    ax_curva.set_title(f"Q(r) Progressivo\nQ = {q_at:.2e}")
    ax_curva.set_xlabel("r (nm)")

# Gerando e salvando
anim = animation.FuncAnimation(fig_anim, update, frames=len(raios_gif), interval=200)
anim.save("nanocateter.gif", writer=PillowWriter(fps=10))
plt.close(fig_anim)
print("GIF e PNG gerados com sucesso e salvos nos arquivos do Colab!")
```
